# Connecting to Databricks

This notebook provides templates for connecting to Databricks from a local environment using two primary methods:
1. **Databricks SQL Connector**: Best for executing SQL queries against a SQL Warehouse.
2. **Databricks Connect**: Best for running Spark code locally while executing on a Databricks cluster.

## 2. Connection via Databricks SQL Connector
Use this method if you only need to run SQL queries.

````python
import os
from databricks import sql

# It is recommended to use environment variables for security
SERVER_HOSTNAME = os.getenv('DATABRICKS_SERVER_HOSTNAME', 'your-instance.cloud.databricks.com')
HTTP_PATH = os.getenv('DATABRICKS_HTTP_PATH', 'your-http-path')
ACCESS_TOKEN = os.getenv('DATABRICKS_TOKEN', 'your-access-token')

try:
    connection = sql.connect(
        server_hostname=SERVER_HOSTNAME,
        http_path=HTTP_PATH,
        access_token=ACCESS_TOKEN
    )
    print("Successfully connected to Databricks SQL Warehouse!")
    
    cursor = connection.cursor()
    cursor.execute("SELECT current_user();")
    print(f"Connected as: {cursor.fetchone()[0]}")
    
    cursor.close()
    connection.close()
except Exception as e:
    print(f"Error connecting to Databricks SQL: {e}")
````

## 3. Connection via Databricks Connect
Use this method to use the PySpark API with a remote Databricks cluster.

````python
from databricks.connect import DatabricksSession

try:
    # DatabricksSession.builder.getOrCreate() looks for configuration in 
    # ~/.databricks/configuration or environment variables
    spark = DatabricksSession.builder.getOrCreate()
    
    print("Successfully connected via Databricks Connect!")
    df = spark.sql("SELECT current_user()").collect()
    print(f"Spark connected as: {df[0][0]}")
except Exception as e:
    print(f"Error connecting via Databricks Connect: {e}")
    print("\nNote: Ensure you have run 'databricks-connect configure' or set the correct environment variables.")
````
